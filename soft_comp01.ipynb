```python
#Part G - Write the complete One-Iteration Program
```

```python
import math

def sigmoid(z):
    return 1/(1+math.exp(-z))
    pass

#Given values
x1=0.5
x2=0.8
target=1.0
eta = 0.5

v11, v12 = 0.4, 0.2
v21, v22 = -0.3, 0.5
w1, w2 = 0.7, -0.4

#Forward Pass
net_h1 = v11*x1+v12*x2
net_h2 = v21*x1+v22*x2
h1 = sigmoid(net_h1)
h2 = sigmoid (net_h2)

net_o = w1*h1 + w2*h2
y = sigmoid(net_o)

#Error
error = 0.5*(target-y)**2

#Output error signal
delta_o = (target - y)*y*(1-y)

#Hidden-to-output updates
delta_w1 = eta*delta_o*h1
delta_w2 = eta*delta_o*h2

w1_new = w1+delta_w1
w2_new = w2+delta_w2

#Hidden error signals
delta_h1 = delta_o * w1 * h1 * (1-h1)
delta_h2 = delta_o * w1 * h2 * (1-h2)

#Input-to-hidden updates
delta_v11 = eta * delta_h1 *x1
delta_v12 = eta * delta_h1 *x2
delta_v21 = eta * delta_h2 *x1
delta_v22 = eta * delta_h2 *x2

v11_new = v11 + delta_v11
v12_new = v12 + delta_v12
v21_new = v21 + delta_v21
v22_new = v22 + delta_v22

print("h1 =", h1) 
print("h2 =", h2) 
print("y  =", y) 
print("Error =", error) 
print("delta_o =", delta_o) 
print("delta_h1 =", delta_h1) 
print("delta_h2 =", delta_h2) 
print("Updated weights:") 
print("w1 =", w1_new) 
print("w2 =", w2_new) 
print("v11 =", v11_new) 
print("v12 =", v12_new) 
print("v21 =", v21_new) 
print("v22 =", v22_new)
```

```python
#Part H - Verify the learning effect
```

```python
net_h1_new = v11_new*x1 + v12_new*x2
net_h2_new = v21_new*x1 + v22_new*x2

h1_new = sigmoid(net_h1_new)
h2_new = sigmoid(net_h2_new)

net_o_new = w1_new*h1_new + w2_new*h2_new
y_new = sigmoid(net_o_new)

error_new = 0.5* (target- y_new)**2

print("New output =", y_new)
print("New error =", error_new)
```
