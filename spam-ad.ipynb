```python
import pandas as pd
import re
import string

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report,precision_score, recall_score, f1_score

RANDOM_STATE = 42
```

```python
path = "sms.txt"
df = pd.read_csv(path, sep="\t", header=None, names=["label", "message"])
df["label_num"] = df["label"].map({"ham": 0, "spam": 1})

print(f"Total messages: {len(df)}")
print(df["label"].value_counts())
df.head()
```

```python
# Data cleansing

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[%s]" % re.escape(string.punctuation), " ", text)
    text = re.sub(r"\d+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
df["clean_message"] = df["message"].apply(clean_text)
df[["message", "clean_message"]].head()
```

```python
# Split the train and test datasets. Vectorize the train and test input features.

X_train_text, X_test_text, y_train, y_test = train_test_split(
    df["clean_message"], df["label_num"],
    test_size=0.2, random_state=RANDOM_STATE, stratify=df["label_num"]
)
vectorizer = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), min_df=2, max_df=0.95)

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

print(f"Train shape: {X_train.shape}")
print(f"Test shape:  {X_test.shape}")
```
