```python
A = np.array([
    [1,1,2],
    [2,1,1],
    [1,2,1]
])
```

```python
B = np.array([
    [1,1,3],
    [3,1,1],
    [1,3,1]
])
```

```python
print("Matrix Addition result:\n",A+B)
print("Matrix Multiplication result:\n",A@B)
print("Matrix Element-wise result:\n",A*B)
print("Transpose Matrix result:\n",A.T)
```

```python
#Implementing Linear Algebra in matrix

print("Matrix:\n",A)
print("Transpose:\n",A.T)
print("Determinant:\n",np.linalg.det(A))
print("Inverse:\n",np.linalg.inv(A))
print("Rank:\n",np.linalg.matrix_rank(A))
print("Trace:\n",np.trace(A))
```
