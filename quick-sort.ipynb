```python
def qs(arr):
    if len(arr) <=1:
        return arr

    pivot = arr[len(arr)//2]
    print(pivot)
    left = [x for x in arr if x < pivot]
    print(left)
    middle = [x for x in arr if x == pivot]
    #print(middle)
    right = [x for x in arr if x > pivot]

    return qs(left) + qs(middle) + qs(right)

#input = [38,27,43,3,9,82,10]
input = [1,2,3,4,5,6,9,7,8]
sorted = qs(input)
print("Sorted :",sorted)
```
