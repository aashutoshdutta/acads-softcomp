```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# 1. Sample Corpus (List of documents)
documents = [
     "The sky is blue and beautiful.",
    "The beautiful blue sky is clear today.",
    "Python is an excellent programming language."
]
print(documents)
```

```python
# 2. Initialize the Vectorizer
# stop_words='english' automatically removes common filler words
vectorizer = TfidfVectorizer(stop_words='english')
print(vectorizer)
```

```python
# 3. Fit and Transform the Documents
tfidf_matrix = vectorizer.fit_transform(documents)
print(tfidf_matrix)
```

```python
# 4. Convert to a DataFrame for clean visualization
feature_names = vectorizer.get_feature_names_out()
print(feature_names)
df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
df
```
